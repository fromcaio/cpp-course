# 05 — Enums and Strong Types 🛡️

## 1. Old C Enums: The Problem

```cpp
enum Color { RED, BLUE };
enum TrafficLight { GREEN, RED }; // ERROR! 'RED' collision.
```

C enums pollute the global namespace and implicitly cast to `int`.


## 2. `enum class`: The Solution

Use `enum class` (Scoped Enums).
1. **Scoped:** accessed via `Color::RED`.
2. **Strongly Typed:** Cannot assign `Color::RED` to an `int` without a cast.

````c++
#include <iostream>

namespace Types {
    enum class Color {
        Red,
        Blue,
        Green
    };

    enum class Status : uint8_t { // You can specify the underlying type!
        Ok = 200,
        NotFound = 404
    };
}

{
    Types::Color c = Types::Color::Red;

    // if (c == 0) // Error! Type safety wins.

    if (c == Types::Color::Red) {
        std::cout << "It is red." << std::endl;
    }
    
    // Casting to int if you really need to
    std::cout << "Code: " << static_cast<int>(Types::Status::NotFound) << std::endl;
}
````

## 3. Casting between `int` and `enum class`

### How to Cast `int` to `enum class`

Use `static_cast<EnumType>(value)` to convert an integer to an enum class type:

```cpp
int code = 404;
Status s = static_cast<Status>(code);
```

### `static_cast` vs C-style cast `(type)`

**`static_cast<type>(value)` (Modern C++):**
- **Explicit and safe**: Clearly shows intent of type conversion in code.
- **Compile-time checked**: The compiler verifies the cast is reasonable.
- **Preferred in C++**: Industry standard and recommended practice.
- **Searchable**: Easy to find all casts in your codebase.

**`(type) value` (C-style cast):**
- **Implicit and dangerous**: Can hide unsafe conversions like `reinterpret_cast`.
- **Performs multiple types of casts**: Could be `static_cast`, `reinterpret_cast`, or `const_cast` all at once.
- **Legacy C code**: Common in old C programs, but discouraged in modern C++.
- **Hard to debug**: Less clear what type conversion is happening.

**Best practice:** Always use `static_cast` in modern C++ instead of C-style casts.

````c++
#include <iostream>

namespace Types {
    enum class Status : uint8_t {
        Ok = 200,
        NotFound = 404,
        ServerError = 500
    };
}

{
    // Example 1: Casting int to enum class using static_cast
    int code = 404;
    Types::Status s = static_cast<Types::Status>(code);
    
    if (s == Types::Status::NotFound) {
        std::cout << "Status is NotFound!" << std::endl;
    }
    
    // Example 2: static_cast in direct comparison
    if (static_cast<Types::Status>(200) == Types::Status::Ok) {
        std::cout << "Status is Ok!" << std::endl;
    }
    
    // Example 3: Casting enum class back to int
    std::cout << "Status code: " << static_cast<int>(Types::Status::ServerError) << std::endl;
    
    // Example 4: C-style cast (NOT recommended)
    // Types::Status s2 = (Types::Status)500; // Works but less safe and clear
}
````
