# 03 — Maps and Sets: Lookups 🔎

## 1. Maps (Dictionaries)

C++ has two main map types:
1. `std::map`: Sorted (Balanced Tree). Keys are always in order. O(log n).
2. `std::unordered_map`: Hash Table. Faster for large data. O(1) avg.

Most C programmers try to implement HashMaps from scratch. Don't.

```c++
#include <map>
#include <unordered_map>
#include <string>
#include <iostream>

{
    // Key: String, Value: Int
    std::map<std::string, int> inventory;

    inventory["Apple"] = 50;
    inventory["Banana"] = 20;
    
    // Insert or update
    inventory["Apple"] += 10; 

    // Access
    std::cout << "Apples: " << inventory["Apple"] << std::endl;

    // Structured Binding (C++17): Unpacking the key/value pair
    std::cout << "--- Inventory List ---" << std::endl;
    for (const auto& [item, quantity] : inventory) {
        std::cout << item << ": " << quantity << std::endl;
    }
}
```

## What is `std::unordered_map`?

`std::unordered_map` is a C++ Standard Library container that stores key-value pairs using a hash table. Unlike std::map, which keeps keys sorted, `std::unordered_map` does not guarantee any order of keys. Instead, it provides average constant-time complexity for insertions, lookups, and deletions.

#### Advantages
- Fast lookups and insertions: Average-case O(1) time complexity.
- No sorting overhead: Keys are not kept in order, making operations faster for large datasets.
- Flexible key types: Any type that is hashable and comparable for equality can be used as a key.

#### When to use
- When you need fast access to elements by key and do not care about the order of keys.
- For large datasets where performance is critical and key ordering is unnecessary.
- When you need a hash table-like structure in C++.

Use `std::unordered_map` when speed is more important than key order. Use `std::map` if you need keys to be sorted.

```c++
#include <unordered_map>
#include <string>
#include <iostream>

{
    // Key: String, Value: Int
    std::unordered_map<std::string, int> ages;

    // Insert or update
    ages["Alice"] = 30;
    ages["Bob"] = 25;
    ages["Alice"] += 1; // Alice has a birthday

    // Access
    std::cout << "Alice's age: " << ages["Alice"] << std::endl;

    // Iteration (order is not guaranteed)
    std::cout << "--- Age List ---" << std::endl;
    for (const auto& [name, age] : ages) {
        std::cout << name << ": " << age << std::endl;
    }
}
```

## 2. Sets (Unique Collections)

`std::set` in C++ is typically implemented using a self-balancing binary search tree, most commonly a Red-Black Tree (which is a type of balanced binary tree, not a B-tree). This ensures that all operations—insert, erase, find—are O(log n).

Difference between set and map
- `std::set` stores only unique keys (no associated value). Each element is itself a key, and the set keeps them sorted.
    - Example: `std::set<int> numbers;`
- `std::map` stores key-value pairs. Each key is unique, and each key maps to a value. The map keeps keys sorted.
    - Example: `std::map<std::string, int> ages;`

#### Summary:

- Use set when you only care about unique keys.
- Use map when you want to associate values with unique keys.

Both use a balanced tree structure for efficient sorted storage and lookup.

```c++
#include <set>

{
    std::set<int> unique_ids;

    unique_ids.insert(10);
    unique_ids.insert(20);
    unique_ids.insert(10); // Duplicate ignored

    std::cout << "Set Size: " << unique_ids.size() << std::endl;

    // Checking existence
    if (unique_ids.count(20)) {
        std::cout << "ID 20 exists" << std::endl;
    }
}
```
